```python
import os
import numpy as np
import pandas as pd
from scipy.io import loadmat
from sklearn.preprocessing import StandardScaler
```

```python
def load_data_from_directory(directory):
    data = []
    labels = []

    # Loop over all subdirectories and files
    for subdir in os.listdir(directory):
        subdir_path = os.path.join(directory, subdir)
        
        if os.path.isdir(subdir_path):
            for file_name in os.listdir(subdir_path):
                if file_name.endswith('.mat'):
                    file_path = os.path.join(subdir_path, file_name)
                    
                    # Load the .mat file
                    mat_data = loadmat(file_path)
                    
                    # Extract the signal data (assuming the signal is stored under the key 'data')
                    signal = mat_data['data'].flatten()  # Flatten if it's a multi-dimensional array
                    
                    # Extract the label (e.g., Fault type and condition from the file name)
                    label = extract_label(file_name)
                    
                    # Append the signal and label to the dataset
                    data.append(signal)
                    labels.append(label)
    
    return np.array(data), np.array(labels)

def extract_label(file_name):
    """
    Extract label based on the file name. 
    For example, '12k_Drive_End_B007_0_118.mat' -> label: 'B007_0'
    """
    # Assuming the label is embedded in the filename, you can extract the fault type and HP.
    if 'normal' in file_name:
        return 'Normal'
    else:
        # Extract fault type and HP (e.g., 'B007_0', 'IR014_1', etc.)
        parts = file_name.split('_')
        return f'{parts[2]}_{parts[3]}'

def preprocess_data(data):
    """
    Preprocess the data: normalization or scaling if necessary.
    """
    scaler = StandardScaler()
    return scaler.fit_transform(data)  # Standardizes each feature (signal) to mean=0, std=1
```

```python
# Define your root directory where the Data folder is located
root_directory = './Data'

# Load data and labels
data, labels = load_data_from_directory(root_directory)

# Preprocess data (optional)
data = preprocess_data(data)

# Convert to a DataFrame for easy handling (optional)
df = pd.DataFrame(data)
df['label'] = labels

# Display the first few rows of the dataset
print(df.head())
```
